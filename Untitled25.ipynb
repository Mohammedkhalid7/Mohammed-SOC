```python
!pip install -q -U google-genai

import time
from google import genai
from google.colab import files, userdata

client = genai.Client(api_key=userdata.get("GEMINI_API_KEY"))

MODELS = [
    "gemini-3.8-flash",
    "gemini-flash-latest",
    "gemini-flash-lite-latest",
    "gemini-pro-latest",
]

logs = """
[2026-10-13T10:00:01Z] Authentication failed for user 'admin' from IP 192.0.2.45
[2026-10-13T10:00:02Z] Authentication failed for user 'admin' from IP 192.0.2.45
[2026-10-13T10:00:03Z] Authentication failed for user 'admin' from IP 192.0.2.45
[2026-10-13T10:00:05Z] Authentication SUCCESS for user 'admin' from IP 192.0.2.45
[2026-10-13T10:00:10Z] User 'admin' created a new IAM policy with Admin privileges.
"""


def analyze_security_logs(log_text, retries_per_model=3):
    prompt = f"""
    You are a cybersecurity analyst in a SOC team. Analyze the following security logs carefully:

    {log_text}

    Provide a detailed report including:
    1. Threat Severity Level (Low, Medium, High, Critical)
    2. Detected Suspicious Activities or Indicators of Compromise (IOCs)
    3. MITRE ATT&CK Mapping: for each detected activity, give the Tactic, Technique name,
       and Technique ID (e.g., T1110 Brute Force), plus the log line that supports it.
    4. Attack Timeline (ordered by time)
    5. Actionable Recommendations and Immediate Incident Response Steps
    """

    for model in MODELS:
        for attempt in range(retries_per_model):
            try:
                response = client.models.generate_content(
                    model=model,
                    contents=prompt,
                )
                print(f"Used model: {model}\n")
                return response.text
            except Exception as e:
                err = str(e)
                if "503" in err or "429" in err:
                    wait = 10 * (attempt + 1)
                    print(f"[{model}] busy, retry in {wait}s...")
                    time.sleep(wait)
                else:
                    print(f"[{model}] failed: {err[:120]}")
                    break
    return "All models failed. Try again in a few minutes."


report = analyze_security_logs(logs)
print(report)

with open("security_report.txt", "w", encoding="utf-8") as f:
    f.write(report)
files.download("security_report.txt")
```
